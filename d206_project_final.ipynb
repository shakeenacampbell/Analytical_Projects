```python
#make a replace outlier function
def rep_outliers_median(df, columns):

    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)
        df[col] = np.where((df[col] < lower_bound) | (df[col] > upper_bound), np.nan, df[col])
        return df
#use it
rep_outliers_median(df,[outlier_columns])
```

```python
#Pca
pca_df= df[['VitD_supp', 'Latitude', 'Longitude', 'Population', 'Children', 'Age', 'Income', 'VitD_levels', 'Doc_visits', 'Full_meals_eaten', 'Initial_days', 'DailyCharge', 'Additional_charges']]
pd.options.mode.copy_on_write = True
pca_df.dropna(inplace=True)
```

```python
#Do Pca  ['Customer_id', 'Interaction', 'UID', 'City', 'State', 'County', 'Area', 'Timezone', 'Job', 'Education', 'Employment', 'Marital', 'Gender', 'ReAdmis', 'Soft_drink', 'Initial_admin', 'Complication_risk', 'Services']
pca_normalized =(pca_df-pca_df.mean())/pca_df.std()
pca_df.head()
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=pca_df.shape[1])
pca.fit(pca_normalized)
medical_data_pca = pd.DataFrame(pca.transform(pca_normalized),     
     columns=['PC1','PC2','PC3','PC4','PC5',"PC6", "PC7", "PC8", "PC9", "PC10", "PC11", "PC12", "PC13"])
```

```python
loadings = pd.DataFrame(pca.components_.T,
     columns=['PC1','PC2','PC3','PC4','PC5','PC6','PC7','PC8',"PC9", "PC10", "PC11", "PC12", "PC13"],
     index=pca_normalized.columns)
df=loadings
loadings
```

```python
#Do Pca plot and download numpy or mat
import seaborn as sns
cov_matrix = np.dot(pca_normalized.T, pca_normalized) / pca_df.shape[0]
eigenvalues = [np.dot(eigenvector.T, np.dot(cov_matrix, eigenvector)) for eigenvector in pca.components_]
plt.plot(pca.explained_variance_ratio_)
plt.xlabel('number of components')
plt.ylabel('explained variance')
plt.show()
```

```python
plt.plot(eigenvalues)
plt.xlabel('number of components')
plt.ylabel('eigenvalues')
plt.axhline(y=1, color='red')
plt.show() 
```
